### **Reading The File**:

#### **`read()`**:

This is used to read the entire file at once, This is very effective when your file is small and you know that the file is small, so loading it into the RAM at once is not harmful.

#### **`readline()`**:

This is used when you want to read the file line by line due to memory problem.

```python
f= open('sample.txt', 'r')
r= f.read()
print(r)
f.close()
```

```python
# Suppose you want to read first 10 charaters only. This is done to get overview of a file.
f= open('sample.txt', 'r')
r= f.read(10)
print(r)
f.close()
```

```python
# readline
f= open('sample.txt', 'r')
print(f.readline(), end= '')# The end='' is used because the readline() and the print both are using the new line so there is a space between two lines
print(f.readline(), end= '')
f.close()
```

```python
# But in real life situation you don't know that how many lines are present in the file so how many times you will print the new line?
# To overcome this we use loop

f= open('sample.txt', 'r')

for line in f:
  print(line, end= '')

f.close()
```

```python
# We can also use while loop
f = open('/content/sample.txt', 'r')

r = f.readline()
while r != '': # Checking for an empty string
    print(r, end='')
    r = f.readline() # Read the next line for the next iteration

f.close()
```

```python
# Let's open in small chunks
with open('bigdata.txt', 'r') as data:
    chunk_size = 6

    # Read the first chunk before starting the loop
    chunk = data.read(chunk_size)

    while len(chunk) > 0:
        print(chunk, end='******')
        # Read the next chunk for the next iteration
        chunk = data.read(chunk_size)
```

#### **tell()**:

This is the method used to know where is the buffer now that is the position of buffer right now.



#### **seek()**:

This is the method used to go anywere inside your file. Suppose you are at 10th character now but you want to skip 5 characters after 10 then you will use `f.seek(15)`.

```python
with open ('sample1.txt', 'r') as f:
  print(f.tell()) # Initially at 0th index
  print(f.read(10)) # reading the first 10 characters
  print(f.tell()) # at 11th index after reading first 10 characters
  f.seek(5) # telling to go at 5th index
  print(f.tell()) # Getting the index after telling to go at 5th index
```

```python
# Seek during write
with open ('sample3.txt', 'w') as f:
  f.write('Hello')
  f.seek(0)
  f.write('X')

with open ('sample3.txt', 'r') as f:
  print(f.read())


```

### **Serialization and Deserialization of Custom Objects**

* We can't serialize directly our custom objects
* To Serialize custom objects we have to tell the json how our object must behave

```python
# let' define a custom class

class Person():

  def __init__(self, name, age, gender):
    self.name= name
    self.age= age
    self.gender= gender

person= Person('Keshav', 25, 'Male')
```

```python
import json
with open ('demo5.json', 'w') as f:
  json.dump(person, f)
```

```python
import json
# Defining the function of behaviour
def default_obj(obj):
  if isinstance(obj, Person):
    return {'name': obj.name, 'age': obj.age, 'gender': obj.gender}
  else:
    raise TypeError('Object of type Person is not JSON serializable')

with open ('demo5.json', 'w') as f:
  json.dump(person, f, default= default_obj)
```

```python
import json
with open ('demo5.json', 'r') as f:
  d= json.load(f)
  print(d)
  print(type(d))
```

## **Pickling**:

Pickling is the process of converting a Python object into a binary format. This way, if you need the object in other files or locations in the future, you can just load it, and it will work exactly the same as it did in the original file.

```python
class Person():

  def __init__(self, name, age, gender):
    self._name= name
    self._age= age
    self._gender= gender


  def get_name(self):
    return self._name

  def get_age(self):
    return self._age

  def get_gender(self):
    return self._gender


person= Person('Keshav', 25, 'Male')
```

```python
import pickle
with open ('personal_info.pkl', 'wb') as f:
  pickle.dump(person, f)
```

```python
import pickle
with open('personal_info.pkl', 'rb') as f:
  i= pickle.load(f)
  print(i.get_name())
  print(i.get_age())
  print(i.get_gender())
```

### Pickle Vs json:

* Pickle lets the user to store data in binary format. JSON lets the user store data in a human readable text format.
